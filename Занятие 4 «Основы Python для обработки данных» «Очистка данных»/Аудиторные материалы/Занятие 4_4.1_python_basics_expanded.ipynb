## 8. Сравнения

Операции сравнения возвращают `True` или `False`.

- `>` — больше;
- `<` — меньше;
- `>=` — больше или равно;
- `<=` — меньше или равно;
- `==` — равно;
- `!=` — не равно.

````python
revenue = 120000
plan = 100000

print(revenue > plan)
print(revenue >= plan)
print(revenue == plan)
print(revenue != plan)
````

## 9. Логические операторы

- `and` — оба условия должны быть истинными;
- `or` — достаточно одного истинного условия;
- `not` — меняет `True` на `False` и наоборот.

````python
revenue = 120000
orders_count = 30

print(revenue >= 100000 and orders_count >= 25)
print(revenue >= 150000 or orders_count >= 25)
print(not revenue < 100000)
````

## 10. Условия `if`, `elif`, `else`

Условие позволяет выполнить разные действия в зависимости от результата проверки.

````python
revenue = 120000
plan = 100000

if revenue >= plan:
    print("План выполнен")
else:
    print("План не выполнен")
````

В Python код внутри условия записывается с отступом в четыре пробела.

### Пример 6. Категория заказа

## 14. Цикл `while`

Цикл `while` работает, пока условие истинно.

Внутри цикла обязательно меняйте переменную, которая участвует в условии.

````python
counter = 1

while counter <= 5:
    print(counter)
    counter = counter + 1
````

### Пример 11. Накопление до порога

````python
daily_revenue = [12000, 15000, 18000, 9000]
target = 40000

total = 0
day = 0

while total < target and day < len(daily_revenue):
    total = total + daily_revenue[day]
    day = day + 1

print("Накопленная выручка:", total)
print("Понадобилось дней:", day)
````

## 15. Короткое знакомство с функциями

Функция объединяет повторяющиеся действия в один именованный блок.

````python
def calculate_revenue(unit_price, quantity):
    revenue = unit_price * quantity
    return revenue

result = calculate_revenue(2500, 4)

print(result)
````

### Пример 12. Функция расчёта суммы со скидкой

````python
def calculate_final_amount(unit_price, quantity, discount_rate=0):
    amount = unit_price * quantity
    final_amount = amount * (1 - discount_rate)
    return final_amount

print(calculate_final_amount(2500, 3))
print(calculate_final_amount(2500, 3, 0.10))
````

## 16. Итоговая мини-практика

Даны заказы интернет-магазина.

Нужно:

1. посчитать общую выручку;
2. найти средний чек;
3. посчитать оплаченные заказы;
4. посчитать крупные заказы больше 10 000 рублей;
5. определить, выполнен ли план 40 000 рублей.

````python
orders = [
    {"order_id": "ORD-001", "amount": 3500, "is_paid": True},
    {"order_id": "ORD-002", "amount": 9200, "is_paid": True},
    {"order_id": "ORD-003", "amount": 12800, "is_paid": False},
    {"order_id": "ORD-004", "amount": 5700, "is_paid": True},
    {"order_id": "ORD-005", "amount": 14600, "is_paid": True}
]

plan = 40000

total_revenue = 0
paid_orders_count = 0
large_orders_count = 0

for order in orders:
    total_revenue = total_revenue + order["amount"]

    if order["is_paid"]:
        paid_orders_count = paid_orders_count + 1

    if order["amount"] > 10000:
        large_orders_count = large_orders_count + 1

average_check = total_revenue / len(orders)

if total_revenue >= plan:
    plan_status = "План выполнен"
else:
    plan_status = "План не выполнен"

print(f"Количество заказов: {len(orders)}")
print(f"Оплаченных заказов: {paid_orders_count}")
print(f"Общая выручка: {total_revenue} руб.")
print(f"Средний чек: {average_check:.2f} руб.")
print(f"Крупных заказов: {large_orders_count}")
print(plan_status)
````
