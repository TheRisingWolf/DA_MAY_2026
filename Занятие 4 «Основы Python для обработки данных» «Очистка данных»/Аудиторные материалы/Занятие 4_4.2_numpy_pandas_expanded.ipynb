## 1. Зачем нужны библиотеки

Python имеет базовые средства, но аналитик часто работает с большими наборами чисел и таблицами.

- **NumPy** — числовые массивы и быстрые вычисления;
- **pandas** — таблицы, фильтрация, группировка и преобразование данных;
- **Matplotlib** — графики.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("NumPy:", np.__version__)
print("pandas:", pd.__version__)
````

## 2. Список Python и массив NumPy

Обычный список может хранить разные типы данных.  
Массив NumPy обычно хранит элементы одного типа и позволяет выполнять операции сразу над всеми значениями.

````python
revenue_list = [12000, 15000, 9000, 18000, 16000]
revenue_array = np.array(revenue_list)

print(revenue_list)
print(revenue_array)
print(type(revenue_list))
print(type(revenue_array))
````

### Свойства массива

- `shape` — размер;
- `ndim` — количество измерений;
- `dtype` — тип элементов;
- `size` — количество элементов.

````python
print("shape:", revenue_array.shape)
print("ndim:", revenue_array.ndim)
print("dtype:", revenue_array.dtype)
print("size:", revenue_array.size)
````

## 3. Создание массивов

````python
array_from_list = np.array([10, 20, 30])
zeros_array = np.zeros(5)
ones_array = np.ones(4)
range_array = np.arange(1, 11, 2)

print(array_from_list)
print(zeros_array)
print(ones_array)
print(range_array)
````

## 4. Индексация и срезы NumPy

````python
values = np.array([100, 200, 300, 400, 500])

print("Первый элемент:", values[0])
print("Последний элемент:", values[-1])
print("Первые три:", values[:3])
print("Со второго по четвёртый:", values[1:4])
````

## 5. Векторные операции

Операция применяется сразу к каждому элементу массива.

````python
prices = np.array([1000, 2000, 3000])

prices_with_growth = prices * 1.10
prices_with_discount = prices * 0.90

print("Исходные цены:", prices)
print("После роста на 10%:", prices_with_growth)
print("После скидки 10%:", prices_with_discount)
````

### Пример 1. Выручка по товарам

````python
prices = np.array([1000, 2500, 4000])
quantities = np.array([5, 3, 2])

revenue = prices * quantities

print("Выручка по товарам:", revenue)
print("Общая выручка:", revenue.sum())
````

## 6. Агрегаты NumPy

Полезные методы:

- `.sum()` — сумма;
- `.mean()` — среднее;
- `.min()` — минимум;
- `.max()` — максимум;
- `.median()` через `np.median()`;
- `.std()` — стандартное отклонение.

````python
revenue = np.array([12000, 15000, 9000, 18000, 16000])

print("Сумма:", revenue.sum())
print("Среднее:", revenue.mean())
print("Медиана:", np.median(revenue))
print("Минимум:", revenue.min())
print("Максимум:", revenue.max())
print("Стандартное отклонение:", revenue.std())
````

## 7. Логические маски NumPy

Маска — массив значений `True` и `False`.

````python
revenue = np.array([12000, 15000, 9000, 18000, 16000])

mask = revenue > 14000

print(mask)
print(revenue[mask])
````

Можно записать короче:

````python
high_revenue = revenue[revenue > 14000]
print(high_revenue)
````

### Практика 1. NumPy

Дан массив:

```python
order_amounts = np.array([2500, 7800, 12500, 4300, 15800])
```

Найдите:

1. общую сумму;
2. среднее;
3. значения больше 10 000;
4. количество значений больше 10 000.

````python
# Сначала попробуйте решить самостоятельно
````

### Решение

````python
order_amounts = np.array([2500, 7800, 12500, 4300, 15800])

total = order_amounts.sum()
average = order_amounts.mean()
large_orders = order_amounts[order_amounts > 10000]
large_orders_count = (order_amounts > 10000).sum()

print("Сумма:", total)
print("Среднее:", average)
print("Крупные заказы:", large_orders)
print("Количество крупных заказов:", large_orders_count)
````

## 8. `Series`

`Series` — одномерный объект pandas. Его можно представить как один столбец таблицы.

````python
cities = pd.Series(
    ["Москва", "Казань", "Самара"],
    name="city"
)

display(cities)
````

````python
print("Имя:", cities.name)
print("Тип данных:", cities.dtype)
print("Количество:", len(cities))
print("Первое значение:", cities.iloc[0])
print("Уникальные значения:", cities.unique())
````

### `Series` с собственным индексом

````python
city_revenue = pd.Series(
    [125000, 87000, 93000],
    index=["Москва", "Казань", "Самара"],
    name="revenue"
)

display(city_revenue)
print("Выручка Москвы:", city_revenue.loc["Москва"])
````

### Несколько условий

В pandas:

- `&` означает `и`;
- `|` означает `или`;
- каждое условие берётся в круглые скобки.

````python
filtered_sales = sales[
    (sales["city"] == "Казань") &
    (sales["quantity"] >= 4)
]

display(filtered_sales)
````

````python
selected_cities = sales[
    (sales["city"] == "Москва") |
    (sales["city"] == "Самара")
]

display(selected_cities)
````

Метод `.isin()` удобен для проверки нескольких значений.

````python
selected_cities = sales[sales["city"].isin(["Москва", "Самара"])]
display(selected_cities)
````

## 18. Группировка `groupby`

Группировка работает по схеме:

1. разделить строки на группы;
2. посчитать показатель внутри каждой группы;
3. собрать итоговую таблицу.

````python
city_revenue = (
    sales
    .groupby("city", as_index=False)["revenue"]
    .sum()
)

display(city_revenue)
````

### Несколько показателей через `agg()`

````python
city_summary = (
    sales
    .groupby("city", as_index=False)
    .agg(
        orders_count=("order_id", "count"),
        total_quantity=("quantity", "sum"),
        total_revenue=("revenue", "sum"),
        average_revenue=("revenue", "mean")
    )
)

display(city_summary)
````

### Группировка по двум признакам

````python
city_product_summary = (
    sales
    .groupby(["city", "product"], as_index=False)
    .agg(
        total_quantity=("quantity", "sum"),
        total_revenue=("revenue", "sum")
    )
)

display(city_product_summary)
````

### Практика 3. Итоги по товарам

Для каждого товара рассчитайте:

- количество строк;
- суммарное количество единиц;
- суммарную выручку.

````python
# Сначала попробуйте решить самостоятельно
````

### Решение

````python
product_summary = (
    sales
    .groupby("product", as_index=False)
    .agg(
        rows_count=("order_id", "count"),
        total_quantity=("quantity", "sum"),
        total_revenue=("revenue", "sum")
    )
)

display(product_summary)
````

## 21. Итоговая мини-практика

На основе таблицы `sales`:

1. найдите общую выручку;
2. найдите город-лидер по выручке;
3. найдите товар-лидер по количеству проданных единиц;
4. выберите заказы с выручкой больше 10 000 рублей;
5. сформулируйте короткий аналитический вывод.

````python
total_revenue = sales["revenue"].sum()

city_summary = (
    sales
    .groupby("city", as_index=False)["revenue"]
    .sum()
    .sort_values("revenue", ascending=False)
)

product_summary = (
    sales
    .groupby("product", as_index=False)["quantity"]
    .sum()
    .sort_values("quantity", ascending=False)
)

high_revenue_orders = sales[sales["revenue"] > 10000]

leader_city = city_summary.iloc[0]["city"]
leader_product = product_summary.iloc[0]["product"]

print(f"Общая выручка: {total_revenue} руб.")
print(f"Город-лидер: {leader_city}")
print(f"Товар-лидер по количеству: {leader_product}")

display(high_revenue_orders)
````

### Пример аналитического вывода

> Наибольшую выручку сформировала Москва. По количеству проданных единиц лидирует мышь.  
> Однако вывод основан на небольшой учебной таблице, поэтому его нельзя автоматически переносить на реальные продажи.
