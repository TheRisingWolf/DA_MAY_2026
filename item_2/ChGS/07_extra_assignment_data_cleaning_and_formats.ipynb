# Типичные ошибки новичков

| Ошибка | Что происходит | Как исправить |
|---|---|---|
| Забыли `.copy()` | Можно случайно изменить исходную таблицу | Перед очисткой делайте `df_clean = df.copy()` |
| Не преобразовали строку в число | Сумма и среднее считаются некорректно или не считаются | Используйте `.astype(int)` или `.astype(float)` |
| Использовали regex там, где нужен простой текст | Можно получить неожиданный результат | Для простого текста используйте `regex=False` |
| Не сбросили индекс после `concat` | Индексы повторяются | Используйте `ignore_index=True` |
| Сохранили CSV с индексом | В файле появилась лишняя колонка | Используйте `index=False` |
| Неправильно указали `orient` для JSON | JSON может получиться неудобным для чтения | Для списка объектов используйте `orient='records'` |
| Не проверили тип даты | Дата осталась строкой | Используйте `pd.to_datetime()` |
