## Цикл `for`

Цикл `for` позволяет поочередно обойти все элементы итерируемой последовательности или итерируемого объекта. Конструкция имеет следующий синтаксис:
```
for <переменная> in <последовательность>:
    <тело цикла>
```

````python
for element in (5, 3,7,1,4):
    print(element)
````

````python
for element in (5, 3,7,1,4):
    print(element + 1)
````

Цикл `while` это наиболее часто встречаемая конструкция в алгоритмах, так как это более гибкая конструкция в сравнении с циклом `for`. Синтаксис данной конструкции такой:

```
while <условие>:
  <тело цикла>

````python
i = 0
while True:
  if i == 4:
    break
  print(i)
  i += 1 # i = i + 1
````

## Наполните множество уникальными квадратами введенных пользователем чисел. Наполнение прекращается в тот момент, когда пользователь вводит 0. В конце выведите полученное множество на экран
