```python
# Создание DataFrame из словаря
data = {
    "Категория": ["A", "A", "B", "B", "C"],
    "Продажи": [10, 15, 20, 25, 30],
    "Город": ["Москва", "СПб", "Москва", "СПб", "Казань"]
}

# создадим датафрейм из словаря, потом выведем датафрейм, потом выведем столбец Продажи Вывдедем первую строку (iloc[0])

# 1 преобразуем словарь (ассоциативный массив {"ключ": "значение"}) в датафрейм (объект pandas) [таблица]
df = pd.DataFrame(data)
```

```python
# Создадим таблицу
data = {
    "Категория": ["A", "A", "B", "B", "C", "C"],
    "Город": ["Москва", "СПб", "Москва", "СПб", "Москва", "СПб"],
    "Продажи": [10, 15, 20, 25, 30, 35]
}
df = pd.DataFrame(data)
print("Исходная таблица:\n", df)
```

```python
df
```

```python
# сводная таблица (povot table) - средние продажи по категориям и городам
pivot = pd.pivot_table(df, values="Продажи", index="Категория", columns="Город", aggfunc="mean")
# сводная таблица
pivot
```

```python
# добавим еще статистику: суммы продаж по городам
pivot_sum = pd.pivot_table(df, values="Продажи", index="Категория", aggfunc="sum")
pivot_sum # Суммарные продажи по городам
```

```python
# matplotlib
x = [i for i in range(6)]
y1 = [i for i in x]
y2 = [i**2 for i in x]
```

```python
fig, ax = plt.subplots()

ax.plot(x, y1, label=('y = x'))
ax.plot(x, y2, label=('y = x**2'))

ax.set_title("main is Figure and Axes")
ax.set_xlabel("ось X")
ax.set_ylabel("ось Y")
ax.grid(True)

ax.annotate("выделим точку", xy=(2, 4), xytext=(3, 5),
            arrowprops=dict(arrowstyle="->"))

plt.show()
```

```python
x = [i for i in range(6)]
y1 = [i for i in x]
y2 = [i**2 for i in x]

# линейные диаграммы с разными стилями
plt.plot(x, y1, color="blue", marker="o", label = "y = x")
plt.plot(x, y2, color="red", linestyle='--', marker="s", label = "y = x**2")

plt.title("Линейная диаграмма")
plt.xlabel("Ось X")
plt.ylabel("Ось Y")
# plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

categoris = ["A", "B", "C", "D", "E"]
values = np.array([5,3,7,8,6]) # значения категорий

# Столбчатая диаграмма (вертикальная)
plt.bar(categoris, values, color='blue', edgecolor="black", label="Данные")

plt.title("Столбчатая диаграмма")
plt.xlabel("Категории")
plt.ylabel("Значения")
plt.legend()
plt.grid(axis="y")
plt.show()
```

```python
categoris = ["A", "B", "C", "D", "E"]
values = np.array([5,3,7,8,6]) # значения категорий

# Столбчатая диаграмма (вертикальная)
plt.barh(categoris, values, color='blue', edgecolor="black", label="Данные")

plt.title("Столбчатая диаграмма")
plt.xlabel("Категории")
plt.ylabel("Значения")
plt.legend()
plt.grid(axis="x")
plt.show()
```
