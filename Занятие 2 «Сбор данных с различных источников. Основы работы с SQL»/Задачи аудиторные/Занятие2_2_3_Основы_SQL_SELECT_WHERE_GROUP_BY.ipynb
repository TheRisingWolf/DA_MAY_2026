# 1. Что такое SQL и база данных

**SQL** — это язык для работы с реляционными базами данных.

С помощью SQL можно:

- создавать таблицы;
- добавлять данные;
- получать данные;
- фильтровать данные;
- сортировать данные;
- группировать данные;
- рассчитывать показатели;
- соединять несколько таблиц.

---

## Основные понятия

| Понятие | Простое объяснение |
|---|---|
| База данных | структурированное хранилище данных |
| Таблица | данные в виде строк и столбцов |
| Строка / запись | один объект, например один пользователь |
| Столбец / поле | характеристика объекта, например имя или возраст |
| Первичный ключ | уникальный идентификатор строки |
| SQL-запрос | команда к базе данных |
| SELECT | команда для получения данных |
| WHERE | условие фильтрации |
| ORDER BY | сортировка |
| GROUP BY | группировка |

# 2. Почему используем SQLite

На этом занятии мы используем **SQLite**.

SQLite — это простая база данных, которая работает прямо внутри Python и Google Colab.

Преимущества SQLite для учебного занятия:

- не нужно устанавливать MySQL;
- не нужен Docker;
- не нужен отдельный сервер;
- все запросы выполняются прямо в ноутбуке;
- базовый SQL-синтаксис похож на MySQL и PostgreSQL.

На следующих этапах эти же принципы можно перенести на MySQL, PostgreSQL и другие СУБД.

# 3. Подготовка рабочей среды

Для занятия нам понадобятся:

1. `sqlite3` — встроенная библиотека Python для работы с SQLite;
2. `pandas` — библиотека для отображения результатов SQL-запросов в виде таблиц.

Также создадим удобную функцию `run_sql()`, чтобы выполнять SQL-запросы и сразу видеть результат в виде DataFrame.

````python
import sqlite3
import pandas as pd
````

````python
# Создаём подключение к базе данных SQLite.
# ':memory:' означает, что база будет создана в оперативной памяти.
# Она будет существовать, пока открыт текущий ноутбук.

connection = sqlite3.connect(':memory:')
cursor = connection.cursor()

print('База данных SQLite создана успешно.')
````

````python
# Вспомогательная функция для выполнения SELECT-запросов.
# Она принимает SQL-запрос в виде текста и возвращает результат как таблицу pandas.

def run_sql(query):
    return pd.read_sql_query(query, connection)
````

# 4. Создание таблицы users

Создадим таблицу пользователей `users`.

В таблице будут поля:

| Поле | Тип данных | Описание |
|---|---|---|
| id | INTEGER | уникальный идентификатор пользователя |
| name | TEXT | имя пользователя |
| email | TEXT | электронная почта |
| age | INTEGER | возраст |
| country | TEXT | страна |
| balance | REAL | баланс |

Поле `id` будет первичным ключом.

````python
# CREATE TABLE — команда создания таблицы.
# PRIMARY KEY означает, что поле id является уникальным идентификатором записи.
# NOT NULL означает, что поле не может быть пустым.

cursor.execute('''
CREATE TABLE users (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT NOT NULL,
    age INTEGER NOT NULL,
    country TEXT NOT NULL,
    balance REAL NOT NULL
);
''')

connection.commit()

print('Таблица users создана успешно.')
````

Для добавления данных используется команда `INSERT INTO`.

Общая логика:

```sql
INSERT INTO table_name (column1, column2, column3)
VALUES (value1, value2, value3);
```

# 8. Псевдонимы столбцов: AS

Иногда названия столбцов в результате запроса нужно сделать более понятными.

Для этого используется `AS`.

Пример:

```sql
SELECT name AS user_name
FROM users;
```

````python
run_sql('''
SELECT
    name AS user_name,
    age AS user_age,
    balance AS user_balance
FROM users;
''')
````

`AS` особенно полезен, когда мы рассчитываем новые показатели или используем агрегатные функции.

# 9. Фильтрация данных: WHERE

`WHERE` используется, чтобы выбрать только те строки, которые соответствуют условию.

Пример:

```sql
SELECT *
FROM users
WHERE age > 26;
```

Этот запрос выбирает пользователей старше 26 лет.

## Важное правило

В SQL текстовые значения пишутся в кавычках:

```sql

```

Числовые значения пишутся без кавычек:

```sql

```

# 10. Несколько условий: AND и OR

Иногда нужно указать не одно, а несколько условий.

Для этого используются:

- `AND` — должны выполниться оба условия;
- `OR` — должно выполниться хотя бы одно условие.

---

## AND

Пример:

```sql
WHERE country = 'RUS' AND balance > 90000
```

Это означает: страна должна быть `RUS` и одновременно баланс должен быть больше 90 000.

````python
run_sql('''
SELECT *
FROM users
WHERE country = 'RUS' AND balance > 90000;
''')
````

## OR

Пример:

```sql
WHERE country = 'USA' OR age < 25
```

Это означает: пользователь либо из `USA`, либо младше 25 лет.

````python
run_sql('''
SELECT *
FROM users
WHERE country = 'USA' OR age < 25;
''')
````

# 11. Диапазоны: BETWEEN

`BETWEEN` используется для отбора значений в диапазоне.

Пример:

```sql
WHERE balance BETWEEN 95000 AND 110000
```

Этот запрос выберет пользователей с балансом от 95 000 до 110 000 включительно.

# 13. Агрегатные функции

Агрегатные функции рассчитывают один показатель по набору строк.

Основные агрегатные функции:

| Функция | Что делает |
|---|---|
| COUNT | считает количество строк |
| AVG | считает среднее значение |
| SUM | считает сумму |
| MIN | находит минимальное значение |
| MAX | находит максимальное значение |

## COUNT: общее количество пользователей

````python
run_sql('''
SELECT COUNT(*) AS users_count
FROM users;
''')
````

## COUNT с условием: количество пользователей из RUS

````python
run_sql('''
SELECT COUNT(*) AS rus_users_count
FROM users
WHERE country = 'RUS';
''')
````

## AVG: средний баланс пользователей

````python
run_sql('''
SELECT AVG(balance) AS avg_balance
FROM users;
''')
````

## SUM: сумма всех балансов

````python
run_sql('''
SELECT SUM(balance) AS total_balance
FROM users;
''')
````

## MIN и MAX: минимальный и максимальный баланс

````python
run_sql('''
SELECT
    MIN(balance) AS min_balance,
    MAX(balance) AS max_balance
FROM users;
''')
````

## Несколько агрегатов в одном запросе

````python
run_sql('''
SELECT
    COUNT(*) AS users_count,
    AVG(age) AS avg_age,
    AVG(balance) AS avg_balance,
    MIN(balance) AS min_balance,
    MAX(balance) AS max_balance
FROM users;
''')
````

# 14. Группировка: GROUP BY

`GROUP BY` используется, когда нужно посчитать показатели не по всей таблице, а отдельно по группам.

Например:

- количество пользователей по странам;
- средний баланс по странам;
- средний возраст по странам.

Обычно `GROUP BY` используется вместе с агрегатными функциями.

## Задание 8. Количество пользователей по странам

````python
run_sql('''
SELECT
    country,
    COUNT(*) AS users_count
FROM users
GROUP BY country;
''')
````

## Задание 9. Средний баланс по странам

````python
run_sql('''
SELECT
    country,
    AVG(balance) AS avg_balance
FROM users
GROUP BY country;
''')
````

## Задание 10. Средний возраст по странам

````python
run_sql('''
SELECT
    country,
    AVG(age) AS avg_age
FROM users
GROUP BY country;
''')
````

## Задание 11. Несколько показателей по странам

````python
run_sql('''
SELECT
    country,
    COUNT(*) AS users_count,
    AVG(age) AS avg_age,
    AVG(balance) AS avg_balance,
    MIN(balance) AS min_balance,
    MAX(balance) AS max_balance
FROM users
GROUP BY country;
''')
````

# 15. Сортировка результата группировки

Результат группировки тоже можно сортировать.

Например, отсортируем страны по среднему балансу по убыванию.

````python
run_sql('''
SELECT
    country,
    AVG(balance) AS avg_balance
FROM users
GROUP BY country
ORDER BY avg_balance DESC;
''')
````

# 17. Частые ошибки в SQL

## Ошибка 1. Забыли кавычки у текста

Неправильно:

```sql
WHERE country = RUS
```

Правильно:

```sql
WHERE country = 'RUS'
```

---

## Ошибка 2. Забыли FROM

Неправильно:

```sql
SELECT name, age;
```

Правильно:

```sql
SELECT name, age
FROM users;
```

---

## Ошибка 3. Неправильный порядок команд

Обычно порядок такой:

```sql
SELECT ...
FROM ...
WHERE ...
GROUP BY ...
ORDER BY ...
LIMIT ...
```

---

## Ошибка 4. Использование агрегата без понимания группировки

Если нужно посчитать средний баланс по странам, недостаточно написать только `AVG(balance)`.  
Нужно добавить `GROUP BY country`.
