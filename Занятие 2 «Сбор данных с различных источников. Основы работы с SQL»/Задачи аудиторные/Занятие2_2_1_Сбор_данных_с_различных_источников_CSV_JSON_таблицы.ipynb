# 3. Часть 1. Работа с CSV-файлом

## Что такое CSV?

**CSV** расшифровывается как **Comma-Separated Values**, то есть «значения, разделённые запятыми».

CSV-файл — это простой текстовый файл, который хранит данные в виде таблицы.

Пример:

```text
Фамилия,Имя,Возраст,Специальность
Иванов,Александр,21,Программирование
Петрова,Анна,20,Экономика
```

Каждая строка — это отдельная запись.  
Каждый столбец отделяется запятой или другим разделителем.

В реальной практике CSV часто используют для выгрузок из CRM, Excel, сайтов, баз данных и аналитических систем.

## Задание 1. Создаём CSV-файл со студентами

На практике данные могут быть уже загружены из внешнего источника.  
Но для учебного занятия мы сначала создадим небольшой CSV-файл прямо в Colab, чтобы все слушатели работали с одинаковыми данными.

````python
# Создадим текст CSV-файла.
# Первая строка — это названия столбцов.
# Остальные строки — это данные о студентах.

students_csv = '''Фамилия,Имя,Возраст,Специальность
Иванов,Александр,21,Программирование
Петрова,Анна,20,Экономика
Кузнецов,Сергей,22,Биология
Сергеев,Дмитрий,21,Программирование
Романова,Елена,20,Экономика
Николаев,Павел,23,Биология
'''

# Сохраним текст в файл students.csv.
# encoding='utf-8' нужен, чтобы корректно сохранялись русские буквы.

with open('students.csv', 'w', encoding='utf-8') as file:
    file.write(students_csv)

print('Файл students.csv создан успешно.')
````

## Задание 2. Загружаем CSV-файл

Теперь загрузим созданный файл с помощью `pd.read_csv()`.

Функция `read_csv()` читает CSV-файл и превращает его в DataFrame.

**DataFrame** — это таблица в pandas.

````python
# Загружаем CSV-файл в DataFrame.

students = pd.read_csv('students.csv')

# Выводим таблицу.
students
````

## Задание 6. Считаем количество студентов

Для подсчёта количества строк можно использовать:

- `len()`;
- `.count()` для непустых значений по столбцам.

В этом занятии используем простой способ через `len()`.

````python
# Количество студентов — это количество строк в таблице.

students_count = len(students)

print('Количество студентов:', students_count)
````

## Задание 7. Считаем средний возраст студентов

Для расчёта среднего значения используется метод `.mean()`.

````python
# Считаем средний возраст по столбцу Возраст.

average_age = students['Возраст'].mean()

print('Средний возраст студентов:', average_age)
````

Чтобы результат выглядел аккуратнее, можно округлить значение.

````python
# Округлим средний возраст до двух знаков после запятой.

average_age = round(students['Возраст'].mean(), 2)

print('Средний возраст студентов:', average_age)
````

## Задание 8. Группировка по специальности

Теперь посчитаем средний возраст отдельно по каждой специальности.

Для этого используется метод `.groupby()`.

Логика такая:

1. сгруппировать строки по специальности;
2. взять столбец `Возраст`;
3. посчитать среднее значение.

````python
# Группируем данные по специальности и считаем средний возраст.

avg_age_by_specialty = students.groupby('Специальность')['Возраст'].mean()

avg_age_by_specialty
````

Результат можно сделать более удобным для чтения, превратив его обратно в таблицу.

````python
# reset_index() превращает результат группировки обратно в обычную таблицу.

avg_age_by_specialty = students.groupby('Специальность')['Возраст'].mean().reset_index()

# Переименуем столбец, чтобы название было понятнее.
avg_age_by_specialty = avg_age_by_specialty.rename(columns={'Возраст': 'Средний возраст'})

avg_age_by_specialty
````

# 4. Часть 2. Работа с JSON

## Что такое JSON?

**JSON** расшифровывается как **JavaScript Object Notation**.

JSON часто используется:

- в API;
- на сайтах;
- в мобильных приложениях;
- в системах обмена данными.

JSON может хранить данные в виде списков и объектов.

Пример одного объекта:

```json
{
  "title": "Запущен новый сервис аналитики данных",
  "author": "Иван Петров",
  "date": "2026-05-10"
}
```

Для аналитика важно уметь преобразовывать JSON в таблицу.

## Задание 9. Создаём JSON-файл с новостями

Создадим небольшой набор данных с новостями.

Каждая новость содержит:

- заголовок;
- автора;
- дату публикации;
- категорию.

````python
# Создаём список словарей.
# Каждый словарь — это одна новость.

news_data = [
    {
        'title': 'Запущен новый сервис аналитики данных',
        'author': 'Иван Петров',
        'date': '2026-05-10',
        'category': 'Технологии'
    },
    {
        'title': 'Компании переходят на отечественные ИТ-решения',
        'author': 'Анна Смирнова',
        'date': '2026-05-12',
        'category': 'Бизнес'
    },
    {
        'title': 'Рынок Data Analytics продолжает расти',
        'author': 'Иван Петров',
        'date': '2026-05-15',
        'category': 'Аналитика'
    },
    {
        'title': 'Новые инструменты визуализации данных становятся популярнее',
        'author': 'Мария Соколова',
        'date': '2026-05-18',
        'category': 'Технологии'
    }
]

# Сохраняем данные в JSON-файл.
# ensure_ascii=False нужен, чтобы русские буквы сохранялись нормально.
# indent=4 делает файл более читаемым.

with open('news.json', 'w', encoding='utf-8') as file:
    json.dump(news_data, file, ensure_ascii=False, indent=4)

print('Файл news.json создан успешно.')
````

## Задание 10. Загружаем JSON-файл

## Задание 16. Создаём таблицу пользователей

Создадим таблицу `users`.

Поля таблицы:

- `id` — уникальный идентификатор пользователя;
- `name` — имя;
- `email` — электронная почта;

- `country` — страна;

````python
# Создаём данные пользователей в виде списка словарей.

users_data = [
    {'id': 1, 'name': 'Ivan', 'email': 'ivan@example.com', 'age': 25, 'country': 'RUS', 'balance': 100000},
    {'id': 2, 'name': 'Gleb', 'email': 'gleb@example.com', 'age': 26, 'country': 'RUS', 'balance': 90000},
    {'id': 3, 'name': 'Sergey', 'email': 'sergey@example.com', 'age': 28, 'country': 'RUS', 'balance': 90000},
    {'id': 4, 'name': 'Andrew', 'email': 'andrew@example.com', 'age': 24, 'country': 'RUS', 'balance': 95000},
    {'id': 5, 'name': 'Bob', 'email': 'bob@example.com', 'age': 25, 'country': 'USA', 'balance': 100000},
    {'id': 6, 'name': 'Tom', 'email': 'tom@example.com', 'age': 29, 'country': 'USA', 'balance': 110000}
]

# Преобразуем список словарей в DataFrame.

users = pd.DataFrame(users_data)

users
````

## Задание 18. Проверяем пропущенные значения

Пропущенные значения могут мешать анализу.

Для проверки используется метод `.isna()`.

Если добавить `.sum()`, мы получим количество пропусков по каждому столбцу.

## Задание 19. Проверяем уникальность идентификаторов

Поле `id` должно быть уникальным.

## Задание 20. Простые аналитические расчёты по пользователям

Рассчитаем несколько показателей:

1. количество пользователей;
2. средний возраст;
3. средний баланс;
4. максимальный баланс;
5. минимальный баланс.

````python
print('Количество пользователей:', len(users))
print('Средний возраст:', round(users['age'].mean(), 2))
print('Средний баланс:', round(users['balance'].mean(), 2))
print('Максимальный баланс:', users['balance'].max())
print('Минимальный баланс:', users['balance'].min())
````

## Задание 21. Группировка пользователей по странам

Посчитаем:

1. количество пользователей по странам;
2. средний баланс по странам;
3. средний возраст по странам.

````python
# Количество пользователей по странам.

users_by_country = users.groupby('country')['id'].count().reset_index()

users_by_country = users_by_country.rename(columns={'id': 'users_count'})

users_by_country
````

````python
# Средний баланс по странам.

avg_balance_by_country = users.groupby('country')['balance'].mean().reset_index()

avg_balance_by_country = avg_balance_by_country.rename(columns={'balance': 'avg_balance'})

avg_balance_by_country
````

````python
# Средний возраст по странам.

avg_age_by_country = users.groupby('country')['age'].mean().reset_index()

avg_age_by_country = avg_age_by_country.rename(columns={'age': 'avg_age'})

avg_age_by_country
````
